**Flatten a list**

```python
# list comprehension method
def flatten1 (myList):
	return [i for j in myList for i in j]

# recursive method
def flatten2 (myList):
	flatList = []
	for item in myList:
		if isinstance(item, list):
			flatList.extend(flatten2(item))
		else:
			flatList.append(item)
	return flatList

list1 = [[0], [1, 2], [3, [4, 5]], [6], [7]]
list2 = [0, [1, 2], [3, [4, 5]], [6], 7]

print("flatten1(list1): ", flatten1(list1)) # works, but only flattens 1 layer of sublists
# print(flatten1(list2)) # error - can't work with list of ints and sublists of ints

print("flatten2(list1): ", flatten2(list1))
print("flatten2(list2): ", flatten2(list2))
```

***Functional Programming***

Ask: use map(), filter() and reduce() to transform a sequence without explicit loops

```python
from functools import reduce

def square_even(nums):
    return list(map(lambda x:x*x, filter(lambda x:x%2==0, nums)))

def product(nums):
    return reduce(lambda a,b:a*b, nums,1)
```

***Custom Iterator***

Ask: Implement a class that follows Python's iterator protocol using **\__iter__\()** and **\__next__\()**

```python
class Countdown:
    def __init__(self, n):
        self.n = n
    def __iter__(self):
        return self
    def __next__(self):
        if self.n<0:
            raise StopIteration
        v = self.n
        self.n-=1
        return v
```

```python
A = Countdown(10)
```

```python
next(iter(A))
```

```python
next(iter(A))
```

***Memorization for Dynamic Programming***

Ask: Using functools.lru_cache to effciently solve a recursive

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def catalan(n):
    if n<=1:
        return 1
    return sum(catalan(i)*catalan(n-1-i) for i in range(n))
```

```python
catalan(10)
```

 Memoization converts many exponential recursive algorithms into polynomial-time solutions by caching previously computed subproblems.

***Design a resuable data processing pipeline***

Ask: Implement a simple pipeline class that composes multiple processing functions into a reusable workflow

```python
from dataclasses import dataclass 
from typing import Callable, Any

@dataclass 
class Pipeline:
    steps: list[Callable[[Any], Any]]
    def __call__(self, x):
        for step in self.steps:
            x = step(x) 
        return x
        
pipeline = Pipeline([ 
    lambda x: x + 1, #function 1
    lambda x: x * 2, #function 2
])

print(pipeline(3))
```

Composable pipelines are widely used in machine learning, ETL systems, and scientific computing. Good software design emphasizes modularity, testability, and reuse.

```python
%%writefile adder/src/api.py

# Return the sum of the strings coerced to floating-point values
def sum(*args: list[str]) -> float:
    def iter(args, sofar: float) -> float:
        if len(args) == 0:
            return sofar
        return iter(args[1:], sofar + float(args[0]))
    return iter(args, 0.0)
```
