### Creating DataFrames from Lists and Dicts
▶ New DataFrame from a **List**  
Pandas automatically assigns numerical row indexes.

```python
data1 = [random.random() for i in range(10000)]
df = pd.DataFrame(data1)
df.head()
```

▶ New DataFrame from a **2D List**  
Column names default to integers. Each subList is a row.

```python
data2 = [[i, random.randint(10,99)] for i in 'ABCDEFGHIJKLMNOPQRSTUVWXYZ']
df = pd.DataFrame(data2)
df.head()
```

▶ New DataFrame from a **Dictionary**  
Dict Keys become column names

```python
data3 = {
    'Model':['T57','T61','T64','T65'],
    'Price':[1.42,1.48,1.73,1.95],
    'Size':['57 inches','61 inches','64 inches','65 inches']}
df = pd.DataFrame(data3)
df
```

Change previous example to use Model number as index.  

```python
df = pd.DataFrame(
    {'Price':data3['Price'],'Size':data3['Size']}, 
    index=data3['Model'])
df
```

▶ New DataFrame from a List of Dictionaries  
Note, missing Length is populated with NaN (not a number).

```python
data4 = [
    {'Ht':63, 'Len':45, 'Wt':2.6}, 
    {'Ht':29, 'Wt':1.7},
    {'Ht':37, 'Len':71, 'Wt':4.2}]
df = pd.DataFrame(data4)
df
```

[Documentation](https://pandas.pydata.org/docs/user_guide/io.html#csv-text-files) of numerous optional parameters.

### Reading & Writing DataFrames to JSON files
[Documentation](https://pandas.pydata.org/docs/user_guide/io.html#csv-text-files) of numerous optional parameters.  

▶ Convert DataFrame to **JSON** string  
No argument - json by columns is default, {column -> {index -> value}}

```python
data4_json = df.to_json()
data4_json
```

Use orient='index' to structure the json by rows, {index -> {column -> value}}.  
You can also strip out the row indices by using orient='records'.

```python
data4_json = df.to_json(orient='index')
data4_json
```

▶ Write to a text file in JSON format.

```python
data4_json = df.to_json('outjson.txt')
data4_json
```

▶ Read same JSON data back in to a DataFrame.

```python
data4 = pd.read_json('outjson.txt')
data4
```

# Pandas - Delete NaN Rows from DataFrame

[StackOverflow Thread](https://stackoverflow.com/questions/13413590/how-to-drop-rows-of-pandas-dataframe-whose-value-in-a-certain-column-is-nan)

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(np.random.randn(12,3))
df
```

```python
df.iloc[::2,0] = np.nan
df.iloc[::3,1] = np.nan
df.iloc[::4,2] = np.nan
df
```

**- Drop all rows that have *any* NaN values**

```python
df.dropna()
```

**- drop only if *all* columns are NaN**

```python
df.dropna(how='all')
```

**- Drop only if NaN in a specific column**

```python
df.dropna(subset=[2])
```

```python
import numpy as np
import pandas as pd
import datetime
import random
```

### Data Formats Supported
Pandas datetime64 can interpret strings, Python datetime, and Numpy datetime64 objects.  
Also note, a list of pd.datetime64 objects are automatically converted to a DatetimeIndex.

```python
a1 = pd.to_datetime([
    '6/1/2020', 
    '6-2-2020',
    datetime.datetime(2020, 6, 3),
    np.datetime64('2020-06-04'),
    np.datetime64('2020-06-05')])
a1
```

Pass in a format argument for custom formatted dates (case matters).

```python
a2 = pd.to_datetime(['2020/14/06', '2020/15/06'], format='%Y/%d/%m')
a2
```

Hours and Minutes too? No problem.

```python
a3 = pd.to_datetime(
    ['2020/6/8 14.05', '2020/6/9 06.45'], format='%Y/%d/%m %H.%M')
a3
```

### Creating a datetime sequence with fixed intervals
freq parameters:  
  D=days, W=weeks, M=months, B=business days, BW=bus weeks, BM=bus months

```python
b1 = [random.random() for i in range(30)]
b2 = pd.date_range('2020-06-01', periods=30, freq='1d')
print(b2)
df = pd.DataFrame({'M':b1}, index=b2)
#df.loc['2020-06-18':]
df[df['M'] > 0.8]
```

```python
b3 = np.random.rand(52)
b4 = pd.date_range('2020-06-01', periods=52, freq='W')
df = pd.DataFrame(b3, index=b4)
df['2020-07-10':'2020-07-28']
```

Alternative to periods, you can give start and stop dates.

```python
b3 = pd.date_range('2020-06-30', '2020-12-31', freq='M')
b3
```

### Dates Index to/from CSV file
Create DataFrame with Dates as Index, Write it to a CSV file, then Read in the CSV data and put the dates as Index

```python
d1 = np.round(6 + 4 * np.random.randn(7), decimals=2)
d2 = np.random.randint(12, 30, size=7)
d3 = pd.Series(pd.date_range('2020-05-29', periods=7, freq='1d'))
df = pd.DataFrame({'alpha':d1, 'beta':d2}, index=d3)

df.to_csv('file01.csv')
df
```

```python
df = pd.read_csv('file01.csv', index_col=0)
print(type(df.index[2]))
df.index = pd.to_datetime(df.index, format='%Y/%m/%d')
print(type(df.index[2]))
df[:'2020/05/31']
```

### Constructing Dates from Multiple Columns
You have Month, Day and Year in separate fields, and need to combine them into a single Datetime field.

```python
yyyy = [random.randint(1995,2020) for i in range(100)]
mm = [random.randint(1,12) for i in range(100)]
dd = [random.randint(1,28) for i in range(100)]
data = [random.random() for i in range(100)]
print(yyyy[5], mm[5], dd[5], data[5])
```

```python
df1 = pd.DataFrame({'year': yyyy,'month': mm, 'day': dd})
df1 = pd.to_datetime(df1) 
df2 = pd.Series(data)
df = pd.concat([df1, df2], axis=1)
df[:5]
```

### Date Arithmetic

```python
appointment = pd.Timestamp('2020-06-04')
appointment.day_name()
```

Uh oh! my appointment is delayed 2 days.  
Here are 3 different ways to add 2 days to the date.

```python
appointment = pd.Timestamp('2020-06-04')
appointment += pd.Timedelta('2 days')
appointment.day_name()
```

```python
appointment = pd.Timestamp('2020-06-04')
appointment += pd.Timedelta(days=2)
appointment.day_name()
```

Date offsets: Day, Hour, Minute, Second, Milli, Micro, Nano 

```python
appointment = pd.Timestamp('2020-06-04')
appointment += pd.offsets.Day(2)
appointment.day_name()
```

NO, it's delayed 2 business days. 

```python
appointment = pd.Timestamp('2020-06-04')
appointment += pd.offsets.BDay(2)
appointment.day_name()
```
